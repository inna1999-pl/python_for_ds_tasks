**4. Індексація**

Створіть Numpy-масив `arr` зі значеннями [15, 22, 35, 44, 58, 67].

Виведіть (кожен пункт оцінюється в 0.5 бала):
- Перший елемент
- Останній елемент
- Елементи з індексами 1 до 3
- Елементи більші за 40

```python
# Я раніше вже використовувала arr як змінну, тому тут буде numpy_arr
numpy_arr = np.array([15, 22, 35, 44, 58, 67])
print(numpy_arr[0])
print(numpy_arr[-1])
print(numpy_arr[1:4])
print(numpy_arr[numpy_arr > 40])
```
