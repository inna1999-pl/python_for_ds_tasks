#### Гіпотеза 2.
Можливо, у Мандалаї більше клієнтів з карткою члена і вони приносять більше доходу. Перевіряємо це за кодом нижче. Гіпотеза не підтвердилася після виконання коду нижче, адже мемберів більше в Янгоні. Але виконаний код показує, що середній чек і для Member, і для Normal більший у Мандалаї. Цей фактор і робить дохід у Мандалаї вищим.

```python
# Список міст і типів клієнтів
cities = ['Yangon', 'Mandalay']
customer_types = ['Member', 'Normal']

rows = []

for city in cities:
    for ctype in customer_types:
        mask = (df['City'] == city) & (df['Customer type'] == ctype)
        count = mask.sum()  # кількість транзакцій
        revenue = df.loc[mask, 'Total'].sum()  # валовий дохід
        avg_check = revenue / count if count != 0 else 0  # середній чек
        rows.append({
            'City': city,
            'Customer type': ctype,
            'Count': count,
            'Revenue': revenue,
            'Avg_check': avg_check
        })

# Створюємо DataFrame
summary_df = pd.DataFrame(rows)

# Вивід таблиці
summary_df
```

### Гіпотеза 3.
Якщо раніше ми визначили, що середня кількість товарів на чек майже однакова в обох містах, а середня ціна за одиницю трохи більша в Мандалаї, ніж у Янгоні, і дійшли висновку, що тут більше справа в тому, що у Мандалаї купують дорожчі товари, то варто припустити, що вибір жителів Мандалаю на користь певних категорій товарів може значно впливати на вищий валовий дохід у Мандалаї. Запустивши код нижче і поглянувши на виведену таблицю, можна побачити, що у Мандалаї категорія Health and beauty має значно вищий середній дохід (17.95) порівняно з Янгоном (12.76).
Отже, розглянувши перевірку всіх 3-х гіпотез, можна зробити висновок, що основна причина більшого валового доходу в Мандалаї — це висока середня вартість продажів у ключовій категорії "Health and beauty".

```python
import pandas as pd

# Унікальні категорії товарів
product_lines = df['Product line'].unique()

rows = []

for product in product_lines:
    # Середній gross income
    avg_yangon = df.loc[(df['City'] == 'Yangon') & (df['Product line'] == product), 'gross income'].mean()
    avg_mandalay = df.loc[(df['City'] == 'Mandalay') & (df['Product line'] == product), 'gross income'].mean()
    
    # Кількість товарів (транзакцій)
    count_yangon = df.loc[(df['City'] == 'Yangon') & (df['Product line'] == product)].shape[0]
    count_mandalay = df.loc[(df['City'] == 'Mandalay') & (df['Product line'] == product)].shape[0]
    
    rows.append({
        'Product line': product,
        'Yangon_avg_gross_income': avg_yangon,
        'Mandalay_avg_gross_income': avg_mandalay,
        'Yangon_count': count_yangon,
        'Mandalay_count': count_mandalay
    })

# Створюємо DataFrame
comparison_df = pd.DataFrame(rows)

# Сортуємо за середнім gross income у Mandalay для наочності
comparison_df = comparison_df.sort_values(by='Mandalay_avg_gross_income', ascending=True).reset_index(drop=True)

# Виводимо таблицю
comparison_df
```

### Гіпотеза 4.
Можливо, тип оплати впливає на більший валовий дохід у Мандалаї. Якщо ми подивимося на результат коду нижче у вигляді таблиці, то побачимо, що в Мандалаї транзакції через Credit card та Cash мають трохи більший середній чек, ніж у Янгоні. При цьому Ewallet приносить більше доходу у Янгоні через більшу кількість покупок і вищий середній чек.
Отже, вищий валовий дохід у Мандалаї також пояснюється тим, що ключові типи оплати (Credit card і Cash) приносять більше доходу, що компенсує менший внесок Ewallet.

```python
import pandas as pd

# Унікальні типи оплати
payment_types = df['Payment'].unique()

rows = []

for payment in payment_types:
    # Yangon
    mask_yangon = (df['City'] == 'Yangon') & (df['Payment'] == payment)
    count_yangon = mask_yangon.sum()
    revenue_yangon = df.loc[mask_yangon, 'gross income'].sum()
    avg_yangon = revenue_yangon / count_yangon if count_yangon != 0 else 0
    
    # Mandalay
    mask_mandalay = (df['City'] == 'Mandalay') & (df['Payment'] == payment)
    count_mandalay = mask_mandalay.sum()
    revenue_mandalay = df.loc[mask_mandalay, 'gross income'].sum()
    avg_mandalay = revenue_mandalay / count_mandalay if count_mandalay != 0 else 0
    
    rows.append({
        'Payment': payment,
        'Yangon_count': count_yangon,
        'Mandalay_count': count_mandalay,
        'Yangon_revenue': revenue_yangon,
        'Mandalay_revenue': revenue_mandalay,
        'Yangon_avg_check': avg_yangon,
        'Mandalay_avg_check': avg_mandalay
    })

# Створюємо DataFrame
payment_comparison_df = pd.DataFrame(rows)

# Сортуємо по Mandalay_revenue для наочності
payment_comparison_df = payment_comparison_df.sort_values(by='Mandalay_revenue', ascending=False).reset_index(drop=True)

payment_comparison_df
```
