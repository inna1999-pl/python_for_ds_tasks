0. Імпортуйте бібліотеку pandas та зчитайте дані у змінну `df` типу `pandas.DataFrame`.

```python
import pandas as pd
```

```python
data_path = '../data/supermarket_sales.csv'
df = pd.read_csv(data_path)
```

1. Дослідимо, який філіал супермаркету ('Branch') є найприбутковішим. Для цього знайдіть сумарний прибуток за кожним філіалом і виявіть, який філіал має найвищий.

```python
df.groupby("Branch")[['Total', 'gross income']].sum().sort_values(by=['gross income', 'Total'], ascending=False).reset_index()
# Філіал C має найвищий Total і Gross
```

2. В якому місті знайходиться філіал з найвищим прибутком? Може в тому місці нам розмітисти ще один магазин.  
Знайдіть відповідь за допомогою функціоналу Pandas.

```python
df.groupby(["Branch", "City"])[['Total', 'gross income']].sum().sort_values(by=['gross income', 'Total'], ascending=False).head(1).reset_index()
# Філіал з найвищим прибутком знаходиться в місті Naypyitaw. Тут можна розмістити ще один магазин.
```

3.1. Створіть зводну таблицю, яка покаже, скільки покупок (інвойсів) було зроблено в кожній з філій (`Branch`) за різними категоріями товарів. Запишіть таблицю в змінну `invoices_by_category` і виведіть змінну на екран.
Ця таблиця допоможе проаналізувати, в якій філії купують найбільше товарів кожної з категорій.

```python
invoices_by_category = pd.pivot_table(df, index='Branch', columns='Product line', values='Invoice ID', aggfunc='count')
invoices_by_category
```

1-ше питання. Більше покупок роблять і на більшу суму купують жінки чи чоловіки?
Виконавши код, бачимо, що жінки роблять трохи більше покупок і на трохи більшу суму.
Відповідь на це питання може бути використана для прийняття бізнес рішень таким чином: робити рекламу з участю жінок і рекламувати товари, що потенційно цікаві жінкам.

```python
df.groupby("Gender").agg({'Invoice ID': 'count', 'Total': 'sum'})
```

2-ге питання. Які категорії товарів приносять більший прибуток від жінок, а які від чоловіків?
Згідно з даними, жінки більше витрачають грошей на Fashion accessories, Food and beverages, Home and lifestyle, Sports and travel. А чоловіки - на Electronic accessories, Health and beauty.
Відповідь на це питання може бути використана для прийняття бізнес рішень таким чином: робити рекламу для чоловіків з фокусом на те, що вони можуть порадувати своїх жінок прикрасами, купувати здорову їжу, щоб довше залишатися молодими, купувати спортивний одяг, щоб було більше мотивації на спорт, більше подорожувати і для цього купувати необхідні речі. А в рекламі для жінок можна заохочувати їх купувати Electronic accessories, щоб полегшити собі життя (робот пилосос, посудомийка і т.д.), а також засоби для здоров'я і краси, можливо, зробити на них знижку.

```python
invoices_by_gender = pd.pivot_table(df, index='Gender', columns='Product line', values='Total', aggfunc='sum')
invoices_by_gender
```
