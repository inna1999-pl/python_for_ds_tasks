````python
import math


def calc_area(r):
    return math.pi * r ** 2


def calc_circumference(r):
    return 2 * math.pi * r


radius = 5

print(
    "Area:",
    calc_area(radius),
    "Circumference:",
    calc_circumference(radius),
)
````

- from math import * → import math (PEP 8 рекомендує уникати wildcard imports, бо вони “засмічують” простір імен)
- calcArea → calc_area (функції пишуться у стилі snake_case)
- calcCircumference → calc_circumference (snake_case замість camelCase)
- def calcArea(r):return ... → розділено на 2 рядки (тіло функції має бути на окремому рядку з відступом)
- додано пробіли:
    - r**2 → r ** 2
    - radius=5 → radius = 5
- print() розбитий на кілька рядків (довгі виклики функцій краще переносити для читабельності)
- кожен логічний блок відокремлений порожнім рядком

Напишіть код для ефективного перетворення стовпців «product_name» і «product_category» на категорійний тип для ефективного використання пам’яті.
Спробуйте вирішити це завдання, мінімізуючи використання пам'яті та зберігаючи читабельність коду. Крім того, виміряйте використання пам’яті до та після перетворення, щоб продемонструвати підвищення ефективності.

Порівняйте кількість памʼяті, яку займає наданий набір даних і новий, створений Вами. В лекції був приклад. В скільки разів менше памʼяті займає новий датафрейм?

````python
import pandas as pd
import numpy as np

np.random.seed(0)

products = ['apple', 'banana', 'cherry', 'dates', 'elderberry']
categories = ['fruit', 'fruit', 'fruit', 'fruit', 'fruit']

N_ROWS = 10**6
df = pd.DataFrame({
    'product_name': np.random.choice(products, N_ROWS),
    'product_category': np.random.choice(categories, N_ROWS),
    'product_price': np.random.uniform(1, 10, N_ROWS),
    'transaction_date': pd.date_range(start='1/1/2000', periods=N_ROWS, freq='h')
})
````

````python
memory_before = df.memory_usage(deep=True).sum()
memory_before
````

````python
df['product_name'] = df['product_name'].astype('category')
df['product_category'] = df['product_category'].astype('category')
````

````python
memory_after = df.memory_usage(deep=True).sum()
memory_after
````

````python
print(f"Пам'ять до: {memory_before / 1024 ** 2:.2f} MB")
print(f"Пам'ять після: {memory_after / 1024 ** 2:.2f} MB")

reduction_factor = memory_before / memory_after
print(f"Пам'ять зменшилась у {reduction_factor:.2f} разів")
````
